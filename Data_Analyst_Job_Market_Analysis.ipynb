```python
import pandas as pd

df_clean = pd.read_excel(
    '/content/indian-job-market-dataset-2025.xlsx'
)

df_clean = df_clean.drop_duplicates().copy()

print(df_clean.shape)
```

```python
print(df_clean[['minimumSalary', 'maximumSalary']].head())

print("\nData types:")
print(df_clean[['minimumSalary', 'maximumSalary']].dtypes)

print("\nMissing salary values:")
print(df_clean[['minimumSalary', 'maximumSalary']].isna().sum())
```

```python
# 1. Select data-related job titles
data_jobs = df_clean[
    df_clean['title'].str.contains(
        r'data|analyst|analytics|business intelligence|scientist',
        case=False,
        na=False
    )
].copy()

# 2. Keep rows with valid, positive salary ranges
salary_data_jobs = data_jobs[
    data_jobs['minimumSalary'].notna() &
    data_jobs['maximumSalary'].notna() &
    (data_jobs['minimumSalary'] > 0) &
    (data_jobs['maximumSalary'] > 0)
].copy()

# 3. Calculate average salaries
print("Data-related job postings:", len(data_jobs))
print("Jobs with valid salary ranges:", len(salary_data_jobs))

print(
    "Average minimum salary:",
    round(salary_data_jobs['minimumSalary'].mean(), 2)
)

print(
    "Average maximum salary:",
    round(salary_data_jobs['maximumSalary'].mean(), 2)
)
```

```python
import matplotlib.pyplot as plt

avg_min = salary_data_jobs['minimumSalary'].mean()
avg_max = salary_data_jobs['maximumSalary'].mean()

plt.figure(figsize=(8, 5))

plt.bar(
    ['Average Minimum Salary', 'Average Maximum Salary'],
    [avg_min, avg_max]
)

plt.title('Average Salary for Data-Related Jobs')
plt.ylabel('Annual Salary (₹)')
plt.ticklabel_format(axis='y', style='plain')
plt.xticks(rotation=10)
plt.tight_layout()
plt.show()
```

```python

import matplotlib.pyplot as plt

# Calculate average salaries by job title
salary_by_title = (
    salary_data_jobs
    .groupby('title')[['minimumSalary', 'maximumSalary']]
    .mean()
)

# Select the 10 job titles with the most postings
top_titles = data_jobs['title'].value_counts().head(10).index

salary_comparison = salary_by_title.loc[
    salary_by_title.index.intersection(top_titles)
]

# Plot the comparison
salary_comparison.plot(kind='bar', figsize=(12, 6))

plt.title('Average Salary Range by Data-Related Job Title')
plt.xlabel('Job Title')
plt.ylabel('Salary (₹)')
plt.xticks(rotation=45, ha='right')
plt.ticklabel_format(axis='y', style='plain')
plt.legend(['Average Minimum Salary', 'Average Maximum Salary'])
plt.tight_layout()
plt.show()
```

```python

# Filter job postings related to data and analytics
data_jobs = df_clean[
    df_clean['title'].str.contains(
        r'data|analyst|analytics|business intelligence|scientist',
        case=False,
        na=False
    )
].copy()

# Extract skills from these job postings
data_skills = (
    data_jobs['tagsAndSkills']
    .dropna()
    .str.split(',')
    .explode()
    .str.strip()
    .str.lower()
)

# Count the most frequently requested skills
top_data_skills = data_skills.value_counts().head(15)

print("Top 15 Skills in Data-Related Job Postings:")
print(top_data_skills)

# Create a chart
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
top_data_skills.sort_values().plot(kind='barh')

plt.title('Top 15 Skills in Data-Related Job Postings')
plt.xlabel('Number of Skill Mentions')
plt.ylabel('Skill')
plt.tight_layout()
plt.show()
```

```python

# Count data-related job postings by location
top_data_locations = data_jobs['location'].value_counts().head(10)

print("Top 10 Locations for Data-Related Jobs:")
print(top_data_locations)

# Create a chart
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

top_data_locations.sort_values().plot(kind='barh')

plt.title('Top 10 Locations for Data-Related Jobs in India')
plt.xlabel('Number of Job Postings')
plt.ylabel('Location')

plt.tight_layout()
plt.show()
```
